```python
# текст запроса

# расчет количества вакансий и работодателей, в зависимости от региона
# анализ таблицы "employers" (список работодателей), таблицы "vacancies" (список вакансий) и таблицы "areas" (список регионов)
# левое соединение по идентификатру города (таблица "areas" и таблица "vacancies")
# левое соединение по идентификатру работодателя (таблица "employers" и таблица "vacancies")
# группировка данных по полю "идентификатору города" и по полю "название города"
# поиск регионов, в которых нет вакансий, найдите тот, в котором наибольшее количество работодателей
query_5_2 = f'''
    SELECT 
        a.name, 
        count(distinct v.id) as vacancies_count, 
        count(distinct e.id) as employers_count
    FROM public.areas a
    JOIN public.vacancies v ON v.area_id = a.id
    JOIN public.employers e ON v.employer_id = e.id
    GROUP BY a.id, a.name

union all
(
    SELECT 
        a.name, 
        0 as vacancies_count, 
        count(distinct e.id) as employers_count
    FROM public.areas a
    JOIN public.employers e ON e.area = a.id
    WHERE NOT EXISTS
    (
        SELECT 1
            FROM public.vacancies v 
            WHERE v.area_id = a.id
    )
    GROUP BY a.id, a.name
    ORDER BY employers_count DESC
    LIMIT 1
)
'''
```

7. Для компании «Яндекс» выведите список регионов-миллионников, в которых представлены вакансии компании, вместе с количеством вакансий в этих регионах. Также добавьте строку Total с общим количеством вакансий компании. Результат отсортируйте по возрастанию количества.

Список городов-милионников надо взять [отсюда](https://ru.wikipedia.org/wiki/%D0%93%D0%BE%D1%80%D0%BE%D0%B4%D0%B0-%D0%BC%D0%B8%D0%BB%D0%BB%D0%B8%D0%BE%D0%BD%D0%B5%D1%80%D1%8B_%D0%A0%D0%BE%D1%81%D1%81%D0%B8%D0%B8). 

Если возникнут трудности с этим задание посмотрите материалы модуля  PYTHON-17. Как получать данные из веб-источников и API. 

```python
# список городов-милионников
# данные берутся из таблицы сайта https://ru.wikipedia.org/wiki/Города-миллионеры_России

import requests
import mwclient
from bs4 import BeautifulSoup

# URL городов-миллионников
url = 'https://ru.wikipedia.org/wiki/Города-миллионеры_России'
# определение user-agent
headers = {"User-Agent": "Mozilla/5.0"}
# GET-запрос к ресурсу и результат ответа сохраняем в переменной response
response = requests.get(url, headers = headers)

# создаём объект BeautifulSoup, указывая html.parser
soup = BeautifulSoup(response.text, 'html.parser')

# Находим таблицу с классом "standard"
table = soup.find('table', {'class': 'standard'})

#сбор данных городов-миллионников
towns_million = []
if table:
    for row in table.find_all('tr'):
        cells = row.find_all('td')
        if len(cells) >= 2:
            city_link = cells[1].find('a')
            if city_link:
                towns_million.append(city_link.get_text(strip = True))
else:
    print('Таблица с городами не найдена')

# получение списка городов-миллионников
towns_million_sql = ", ".join([f"'{i}'" for i in towns_million])
print(towns_million_sql)
```

```python
# текст запроса

# подсчет количества вакансий для компании "Яндекс" в городах-миллионниках
# анализ таблицы "employers" (список работодателей), таблицы "vacancies" (список вакансий) и таблицы "areas" (список регионов)
# левое соединение по номеру работодателя (таблица "employers" и таблица "vacancies")
# левое соединение по идентификатору города (таблица "employers" и таблица "areas")
# условие отбора данных: поле "название работодателя" содержит "Яндекс" и поле "название города" содержится в списке городов-миллионников
query_5_7 = f'''
    WITH v_vacancies_million as
    (
        SELECT 
            a.name as area_name, 
            count(*) as count
        FROM public.employers e
        JOIN public.vacancies v ON v.employer_id = e.id
        JOIN public.areas a ON a.id = v.area_id
        WHERE 
            e.name ILIKE 'Яндекс' AND a.name IN ({towns_million_sql})
        GROUP BY e.id, a.id, a.name
    )

    (SELECT 
        vm.area_name as "город миллионник", 
        vm.count as "кол-во вакансий"
    FROM v_vacancies_million vm 
    ORDER BY count DESC)
    UNION ALL
    (SELECT 
        'Total' as area_name, 
        sum(vm.count) as count 
    FROM v_vacancies_million vm 
    ORDER BY count DESC)
'''
```

```python
# результат запроса
df = pd.read_sql_query(query_5_7, connection)
df
```

6. Напишите запрос, позволяющий вычислить, какую зарплату для DS в **среднем** указывают для каждого типа требуемого опыта (уникальное значение из поля *experience*). 

При решении задачи примите во внимание следующее:
1. Рассматриваем только вакансии, у которых заполнено хотя бы одно из двух полей с зарплатой.
2. Если заполнены оба поля с зарплатой, то считаем зарплату по каждой вакансии как сумму двух полей, делённую на 2. Если заполнено только одно из полей, то его и считаем зарплатой по вакансии.
3. Если в расчётах участвует null, в результате он тоже даст null (посмотрите, что возвращает запрос select 1 + null). Чтобы избежать этой ситуацию, мы воспользуемся функцией [coalesce](https://postgrespro.ru/docs/postgresql/9.5/functions-conditional#functions-coalesce-nvl-ifnull), которая заменит null на значение, которое мы передадим. Например, посмотрите, что возвращает запрос `select 1 + coalesce(null, 0)`

Выясните, на какую зарплату в среднем может рассчитывать дата-сайентист с опытом работы от 3 до 6 лет. Результат округлите до целого числа. 

```python
# текст запроса

# подсчет средней зарплаты для каждого типа требуемого опыта
# анализ таблицы "vacancies" (список вакансий)
# в наименовании которых содержится слово: 'data scientist', 'data science', 'исследователь данных', 'machine learning', 'машинн%обучен', 'ML'
# в наименовании которых не содержится слово: 'HTML'
# требуемый опыт содержит значение от 3 до 6 лет
query_6_6 = f'''
    SELECT 
        round
        (
            avg
            ( 
                case
                    WHEN v.salary_from IS NOT NULL AND v.salary_to IS NOT NULL THEN (v.salary_from + v.salary_to) / 2
                    WHEN v.salary_from IS NOT NULL THEN v.salary_from
                    ELSE v.salary_to
                end
            )
            , 0
        )::int as salary_avg
    FROM public.vacancies v
    WHERE
        (
            1=0
            OR name ILIKE '%data scientist%'
            OR name ILIKE '%data science%'
            OR name ILIKE '%исследователь данных%'
            OR name LIKE '%ML%'
            OR name ILIKE '%machine learning%'
            OR name ILIKE '%машинн%обучен%'
        )
        AND name NOT ILIKE '%HTML%'
        AND (v.salary_from IS NOT NULL OR v.salary_to IS NOT NULL)
        AND v.experience = 'От 3 до 6 лет'
'''
```

```python
# результат запроса
df = pd.read_sql_query(query_6_6, connection)
df
```

```python
# Дополнительное исследование 3 (средняя заработная плата по регионам)

# выборка 10 самых популярных регионов, по размеру средней заработной платы
# анализ таблицы "vacancies" (список вакансий) и таблицы "areas" (список регионов)
# левое соединение по идентификатору города
query_area = f'''
    SELECT a.name AS region,
    ROUND(AVG(COALESCE((v.salary_from + v.salary_to) / 2, v.salary_from, v.salary_to)), 0) AS avg_salary
    FROM vacancies v
    JOIN areas a ON v.area_id = a.id
    WHERE v.salary_from IS NOT NULL OR v.salary_to IS NOT NULL
    GROUP BY a.name
    ORDER BY avg_salary DESC
    LIMIT 10
'''

# запрос (10 самых популярных регионов, по размеру средней заработной платы)
df_areas_top_zp = pd.read_sql_query(query_area, connection)

# график: столбчатая диграмма
fig = px.bar(
    data_frame = df_areas_top_zp,
    #x = 'region',   # ось абсцисс (x), регионы
    #y = 'avg_salary', # ось ординат (y), средняя заработная плата (руб.)
    x = 'avg_salary',   # ось абсцисс (x), регионы
    y = 'region', # ось ординат (y), средняя заработная плата (руб.)
    text_auto = True, # отображение надписей, на колонках гистограммы
    title = 'Средняя заработная плата по регионам, руб.', # заголовок
    width = 1000,    # ширина графика
    height = 500,    # высота графика
    text = 'region',  # вывод регионы
    color = 'region', # расраска по регионам
)

# отображение графика (визуализация)
fig.update_layout(
    bargap = 0.1, # ширина вертикальной разделительной полосы
    xaxis_title = 'Регионы', # подпись оси абсции (x)
    yaxis_title = 'Средняя заработная плата, руб.', # подпись оси ординат (y)
)

fig.show()
```

```python
# Дополнительное исследование 4 (средняя заработная плата по вакансиям)

# выборка 10 самых популярных вакансий, по размеру средней заработной платы
# анализ таблицы "vacancies" (список вакансий)
query_avg_salary = f'''
    SELECT name AS vacancies, ROUND(AVG(COALESCE((salary_from + salary_to) / 2, salary_from, salary_to)), 0) AS avg_salary
    FROM vacancies v
    WHERE v.salary_from IS NOT NULL OR v.salary_to IS NOT NULL
    GROUP BY 1
    ORDER BY avg_salary DESC
    LIMIT 10
'''

# запрос
df = pd.read_sql_query(query_avg_salary, connection)

# график: столбчатая диграмма
fig = px.bar(
    data_frame = df, # источник данных
    x = 'avg_salary', # ось абсцисс (x), средняя заработная плата
    y = 'vacancies',  # ось ординат (y), название вакансий
    title = 'Средняя заработная плата по вакансиям, руб.', # заголовок графика
    #color = 'count', # раскраска
    #labels = {'count': 'количество'}, # легенда
    text_auto = True # подписи величин
)

fig.update_layout(
    xaxis_title = 'Средняя заработная плата', # ось абсцисс (x)
    yaxis_title = 'Название вакансии', # ось ординат (y)
    title_x = 0.5,
    showlegend = False
)

fig.show()
```
